```python
import pandas as pd

df = pd.read_csv("customer_shopping_behavior.csv")

print(df.shape)
print(df.head())
print(df.columns.tolist())
```

```python
#Make the column names SQL-friendly
df.columns = [
    "customer_id",
    "age",
    "gender",
    "item_purchased",
    "category",
    "purchase_amount_usd",
    "location",
    "size",
    "color",
    "season",
    "review_rating",
    "subscription_status",
    "shipping_type",
    "discount_applied",
    "promo_code_used",
    "previous_purchases",
    "payment_method",
    "frequency_of_purchases"
]
```

```python
#Create the data for Supabase
data = df.to_dict(orient="records")

print(len(data))
```

```python
#Check the problematic row
nan_count = 0

for row in data:
    for key, value in row.items():
        if pd.isna(value):
            nan_count += 1

print("Remaining NaN:", nan_count)
```

```python
# Find which columns contain the 37 missing values
print(df.isna().sum())
```

```python
import numpy as np

data = df.to_dict(orient="records")

for row in data:
    for key, value in row.items():
        if isinstance(value, (float, np.floating)) and np.isnan(value):
            row[key] = None
```

```python
nan_count = sum(
    isinstance(value, (float, np.floating)) and np.isnan(value)
    for row in data
    for value in row.values()
)

print("Remaining NaN:", nan_count)
```

```python
#Upload to supabase
import requests

url = "https://mbugnxztgvvftmmasjyh.supabase.co/rest/v1/customer_behavior"

headers = {
    "apikey": "sb_publishable_jMVFzkRk1HEveLePH43leA_zaVFDUnv",
    "Content-Type": "application/json",
    "Prefer": "return=minimal"
}

batch_size = 500

for i in range(0, len(data), batch_size):
    batch = data[i:i + batch_size]

    response = requests.post(
        url,
        headers=headers,
        json=batch
    )

    print(
        f"Rows {i + 1}-{min(i + batch_size, len(data))}: "
        f"{response.status_code}"
    )

    if response.status_code not in [200, 201]:
        print(response.text)
        break
```
